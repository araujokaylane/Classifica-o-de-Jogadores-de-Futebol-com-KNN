# Classificação de Jogadores com KNN
Notebook simples de exemplo.

```
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

dados = pd.DataFrame({
    'velocidade': [7, 9, 5, 6],
    'passe': [8, 6, 7, 5],
    'classe': ['Atacante', 'Atacante', 'Meia', 'Zagueiro']
})

X = dados[['velocidade', 'passe']]
y = dados['classe']

modelo = KNeighborsClassifier(n_neighbors=3)
modelo.fit(X, y)

modelo.predict([[8, 7]])
```
